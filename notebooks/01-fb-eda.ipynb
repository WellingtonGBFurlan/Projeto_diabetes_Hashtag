```python
# fazendo um mapa de cores

fig, ax = plt.subplots(figsize=(7,7))

sns.heatmap(df.corr(), annot=True, fmt=".2f", ax=ax, cmap="coolwarm")

plt.show()
```

```python
matriz = np.triu(df.corr())

fig, ax = plt.subplots(figsize=(7,7))

sns.heatmap(df.corr(), mask=matriz, annot=True, fmt=".2f", ax=ax, cmap="coolwarm")

plt.show()
```

```python
# verificando se todos os valores do df podem ser escritos como int

colunas_valores_inteiros = []

for coluna in df.columns:
    if df[coluna].apply(lambda x: x.is_integer()).all():
        colunas_valores_inteiros.append(coluna)

colunas_valores_inteiros
```

```python
# todos os vlaores dessas colunas podem ser escritos como int 
```

```python
#convertendo para int

df[colunas_valores_inteiros] = df[colunas_valores_inteiros].astype(int)

df.info()
```

```python
# convertendo de int64 para int8 paa ocupar menos espaço e otimizar o código

df[colunas_valores_inteiros] = df[colunas_valores_inteiros].apply(
    pd.to_numeric,
    downcast="integer"
)
df.info()
```

```python
# pegando todas as colunas que não são inteiros

df.columns.difference(colunas_valores_inteiros)
```

```python
# convertendo de float64 para float32 paa ocupar menos espaço e otimizar o código

df[df.columns.difference(colunas_valores_inteiros)] = df[df.columns.difference(colunas_valores_inteiros)
].apply(pd.to_numeric, downcast="float")

df.info()
```
